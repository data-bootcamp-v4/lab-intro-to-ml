```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
```

```python
spaceship = pd.read_csv("https://raw.githubusercontent.com/data-bootcamp-v4/data/main/spaceship_titanic.csv")
spaceship.head()
```

```python
print(spaceship.shape)
print(spaceship.info())
```

```python
spaceship.dtypes
```

```python
spaceship.isnull().sum()
```

```python
spaceship_clean = spaceship.dropna()
```

```python
spaceship_clean.shape
```

```python
spaceship_numbers = spaceship_clean.select_dtypes(include="number")

spaceship_numbers.shape
```

```python
spaceship_numbers.head()
```

```python
x = spaceship_numbers.select_dtypes(include="number") 
y = spaceship_clean["Transported"]   
```

```python
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=0)

print("Training set shape:", x_train.shape, y_train.shape)
print("Testing set shape:", x_test.shape, y_test.shape)
```

```python
from sklearn.neighbors import KNeighborsClassifier
```

```python
knn = KNeighborsClassifier()
```

```python
knn.fit(x_train, y_train)
```

```python
pred = knn.predict(x_test)
pred
```

```python
knn.score(x_test, y_test)
```

```python
knn.score(x_train, y_train)
```
